```
import spacy

nlp_sm = spacy.load("en_core_web_sm")
nlp_lg = spacy.load("en_core_web_lg")
nlp_trf = spacy.load("en_core_web_trf")

doc_sm = nlp_sm(article.text)
doc_lg = nlp_lg(article.text)
doc_trf = nlp_trf(article.text)

def counter_ner_types(doc):
    ner_types = {}

    for ent in doc.ents:
        if ent.label_ in ner_types:
            ner_types[ent.label_] += 1
        else:
            ner_types[ent.label_] = 1

    return ner_types

print('Small model:', counter_ner_types(doc_sm))
print('Large model:', counter_ner_types(doc_lg))
print('Transformer model:', counter_ner_types(doc_trf))




```

## **Part 3: Types and tokens**

To practice the theoretical contents, this part will analyze the types and tokens of the loaded article. The Type-Token Ratio is interpreted as a value between 0 and 1, where higher values indicate greater lexical diversity.

In this exercise you will also encounter the `Counter` library that is mainly used to count the frequency of elements in an iterable, such as lists, strings, and tuples.


👋 ⚒ Exercise 6: Complete the next code cell to calculate the number of tokens and types as well as the Type-Token Ratio (TTR), which is number of types / number of tokens. At which unique token frequency do the words get semantically interesting? How does the TTR value change when you cut the length of the article in half?

```
import re
from collections import Counter

# Lowercasing and keeping only letters and apostrophes
tokens = re.findall(r"[a-z']+", article.text.lower())

# Counting how often each unique token appears in the text
counter = Counter(tokens)
print(f"Count of unique tokens: {counter}")

n_tokens = len(tokens)  # TODO: number of tokens
n_types = len(counter)   # TODO: number of types
ttr = n_types/n_tokens       # TODO: type-token ratio

print("tokens:", n_tokens)
print("types: ", n_types)
print("TTR:   ", ttr)
print(counter.most_common(20))



# article cut in half-code

half_tokens = tokens[:len(tokens)//2]

half_counter = Counter(half_tokens)

half_n_tokens = len(half_tokens)
half_n_types = len(half_counter)
half_ttr = half_n_types/half_n_tokens

print("Half tokens:", half_n_tokens)
print("Half types:", half_n_types)
print("Half TTR:", half_ttr)
```

The most frequent tokens are mainly function words and are not very semantically informative. The words become more interesting at lower frequencies where content words related to the topic such as "AI" and "discrimination" appear.
